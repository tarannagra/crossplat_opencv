# Image Transformations Demo Notebook

```
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Load Image

```
img=cv.imread('sample.jpg',0); plt.imshow(img,cmap='gray'); plt.axis('off')
```

## Translation

```
M=np.float32([[1,0,50],[0,1,30]]); out=cv.warpAffine(img,M,(img.shape[1],img.shape[0])); plt.imshow(out,cmap="gray"); plt.axis("off")
```

## Rotation

```
c=(img.shape[1]//2,img.shape[0]//2); M=cv.getRotationMatrix2D(c,45,1); out=cv.warpAffine(img,M,(img.shape[1],img.shape[0])); plt.imshow(out,cmap="gray"); plt.axis("off")
```

## Reflection

```
out=cv.flip(img,1); plt.imshow(out,cmap="gray"); plt.axis("off")
```

## Interpolation

```
r1=cv.resize(img,None,fx=2,fy=2,interpolation=cv.INTER_NEAREST); r2=cv.resize(img,None,fx=2,fy=2,interpolation=cv.INTER_LINEAR); r3=cv.resize(img,None,fx=2,fy=2,interpolation=cv.INTER_CUBIC); fig,ax=plt.subplots(1,3); [ (ax[i].imshow(im,cmap="gray"),ax[i].axis("off")) for i,im in enumerate([r1,r2,r3]) ]
```

## Fourier Transform

```
f=np.fft.fftshift(np.fft.fft2(img)); mag=20*np.log(np.abs(f)+1); plt.imshow(mag,cmap="gray"); plt.axis("off")
```

## Frequency Filtering

```
rows,cols=img.shape; crow,ccol=rows//2,cols//2; mask_lp=np.zeros_like(img); mask_lp[crow-30:crow+30,ccol-30:ccol+30]=1; mask_hp=1-mask_lp; f_lp=f*mask_lp; f_hp=f*mask_hp; lp=np.abs(np.fft.ifft2(np.fft.ifftshift(f_lp))); hp=np.abs(np.fft.ifft2(np.fft.ifftshift(f_hp))); fig,ax=plt.subplots(1,2); ax[0].imshow(lp,cmap="gray"); ax[0].axis("off"); ax[1].imshow(hp,cmap="gray"); ax[1].axis("off")
```
